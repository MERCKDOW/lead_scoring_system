```python
from kfp import dsl


@dsl.container_component
def gpu_check_component():
    return dsl.ContainerSpec(
        image='us-central1-docker.pkg.dev/expanded-nebula-754/chris-containers/gpu-check:latest',
        command=['python3', 'torch_check.py'],
    )


```

```python
@dsl.pipeline(name='gpu-check-pipeline')
def pipeline():
    step1 = gpu_check_component()
    step1.set_gpu_limit(1)
    step1.set_accelerator_type('NVIDIA_TESLA_T4')
    step1.set_cpu_limit('4')
    step1.set_memory_limit('16G')

```

```python
from kfp.v2 import compiler

compiler.Compiler().compile(
    pipeline_func=pipeline,
    package_path='gpu_check_pipeline.json'
)
```

```python

from google.cloud import aiplatform

aiplatform.init(project='expanded-nebula-754', location='us-central1')

job = aiplatform.PipelineJob(
    display_name='gpu-check-pipeline',
    template_path='gpu_check_pipeline.json',
    pipeline_root='gs://cdow-central1/pipeline-root/',
)

job.run()
```
